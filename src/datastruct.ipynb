```python
s1=set([1,2,2,2,2,2,3,4,4,4,4,4,5,6,6,6,6,67,8,9])
s2={10,20,30,30,40,40,50,60,60,60,70,80,90,100,8,9}

print(s1|s2) #union
print(s1-s2) #difference
print(s1&s2) #intersection
```

```python
s1.difference_update(s2)
print(s1)
```

```python
print(s1)
```

```python
s1.isdisjoint(s2)

s3=set(['a','b','c'])

rs = s1.isdisjoint(s3);
print(rs)
s3.pop()
print(s3)
```
