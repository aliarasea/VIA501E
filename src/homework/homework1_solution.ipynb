```python
def problem5(moviefilename,userfilename,ratingsilename,outputfilename="merged.dat"):
    """ we need to first form a dictionary where movie ids are key and,
        movie titles,movie genre and movie years are stored as values.
        Afer that we we will form a dictionary for user data.
        keys will be user id's and values will be user gender,age,occupation and zipcodes.
        In the last part, we will go through each line of ratings data and for each movie id
        and user id we will get correspoding information from the dictionaries we built.
        """
    import re
    moviefile=open(moviefilename,encoding="latin-1")
    movies=moviefile.readlines()
    moviedict={}
    pattern=r"\((\d+)\)" # to match year in title, note that year in (year) format
    for movie in movies:
        #first we need to strip whitespaces and split using "::"
        movie_id,movie_title,movie_genre=movie.strip().split("::")
        movie_year=re.findall(pattern,movie_title)[0]
        movie_title=re.sub(pattern,"",movie_title) # substitutes the year with an empty string
        moviedict[movie_id]=[movie_title,movie_genre,movie_year]
        
    #now we will create a dictionary for user information
    userdict={}
    
    userfile=open(userfilename)
    users=userfile.readlines()
    for user in users:
        user_id,user_gender,user_age,user_occupation,user_zipcode=user.strip().split("::")
        userdict[user_id]=[user_gender,user_age,user_occupation,user_zipcode]
        
    
    #now we will go through each line of ratings.dat file. in each line using userid and
    #movie id will get corresponding informations from the dictionaries we just created
    
    ratingsfile=open(ratingsilename)
    ratings=ratingsfile.readlines()
    #lets open a file for merged output we will writes merged file as we go through ratings lines
    outputfile=open(outputfilename,"w+")
    for line in ratings:
        user_id,movie_id,rating,timestamp=line.strip().split("::")
        output="::".join([user_id,movie_id,rating,timestamp,userdict[user_id][0],
            userdict[user_id][1],userdict[user_id][2],
            userdict[user_id][3],moviedict[movie_id][0],moviedict[movie_id][1],
            moviedict[movie_id][2]])
        outputfile.write(output+"\n")
        
    outputfile.close()

```
